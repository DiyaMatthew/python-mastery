```python
import pandas as pd
```

```python
people = { "first_name" : ["Smith","Rowan","Alfie"],
           "last_name" : ["Will","John","Haze"],
           "age" : ["34", "45","23"],
           "email" : ["smithwill@gmail.com", "rowanjohn@gmail.com", "alfiehaze@gmail.com"]}
    
```

```python
people_df = pd.DataFrame(people)
```

```python
people_df
```

```python
people_df
```

```python

# removing columns

people_df.drop(columns=["first_name","last_name"])
```

```python

# splitting the values in a column and add it to new column

people_df['full_name'].str.split(' ', expand=True) # expand the list values in different columns
```

```python

# to name the columns

people_df[["first","last"]] = people_df["full_name"].str.split(" ", expand = True)
```
