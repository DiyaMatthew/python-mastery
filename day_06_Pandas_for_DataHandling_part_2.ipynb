```python

# Scenario 

You have a very wide DataFrame (172 columns) and you want to:

Find if any column contains programming-language values
and return the column name(s)

🧠 Real-life idea

Imagine a big Excel sheet with 172 columns.
You don’t know which column is “Skills / Technology / Language”, but you know the values look like:

"Python", "Java", "C++", "JavaScript"

So we scan each column’s values and see if any of them contain programming languages.
```

```python
languages = ["python"]
```

```python

# Search only for python

python_columns = []

for col in df.columns:
    if df[col].astype(str).str.contains("python", case=False, na=False).any():
        python_columns.append(col)

python_columns
```

```python

# Search for any programming language

cols_with_languages = []

for col in df.columns:
    # convert column to string & lowercase (safe for mixed data)
    values = df[col].astype(str).str.lower()
    
    if values.str.contains('|'.join(languages), na=False).any():
        cols_with_languages.append(col)

cols_with_languages
```

```python

# If you want to see SAMPLE values from those columns

for col in cols_with_languages:
    print(f"\nColumn: {col}")
    print(df[col].dropna().unique()[:10])
```


# str.contains()

str.contains() is a pandas string method used to check whether a piece of text exists inside each value of a column.

👉 It returns True or False for every row.

👉 It checks whether a text appears inside each value of a column.

Think of it like:

“Does this cell contain this word or not?”

🧠 Real-life analogy

Imagine a list of messages:

Message
"I love Python"
"Java developer"
"Python & SQL"

If you ask:

“Which messages contain the word Python?”

That’s exactly what str.contains() does.

🔍 What it actually does

For each row:

Looks inside the text

Checks if "Python" appears anywhere

Returns True if yes, False if no

⚠️ Handle missing values (important)

df["skills"].str.contains("Python", na=False)

Without na=False, missing values (NaN) cause issues.

🔡 Case-insensitive search

df["skills"].str.contains("python", case=False, na=False)


Matches:

Python

PYTHON
