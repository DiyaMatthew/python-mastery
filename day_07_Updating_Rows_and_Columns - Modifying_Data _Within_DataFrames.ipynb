```python
# Updating Rows and columns and Modifying Data within the DataFrames
```

```python
import pandas as pd
```

```python
people = { "first_name" : ["Smith","Rowan","Alfie"],
           "last_name" : ["Will","John","Haze"],
           "age" : ["34", "45","23"],
           "email" : ["smithwill@gmail.com", "rowanjohn@gmail.com", "alfiehaze@gmail.com"]}
    
```

```python
people_df = pd.DataFrame(people)
people_df
```

```python

# Updating columns

people_df.columns = [x.upper() for x in people_df.columns]
```

```python
people_df
```

```python
people_df.columns = people_df.columns.str.replace('_',' ')
```

```python
people_df
```

```python
people_df.columns = people_df.columns.str.replace(' ', '_')
```

```python
people_df
```

```python
people_df.columns = [x.lower() for x in people_df.columns]
```

```python
people_df
```

```python

# rename a particular column name

people_df.rename(columns = {"first_name" : "first", "last_name" : "last"}, inplace=True)
```
